```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/raw/garments_worker_productivity.csv")
print(f"Filas: {df.shape[0]}  |  Columnas: {df.shape[1]}")
df.head()
```

## 1. Verificación del esquema y calidad de datos

```python
# La columna department viene con inconsistencias de escritura
print(df["department"].value_counts())
```

## 2. Variable objetivo derivada: ¿el equipo cumplió su meta?

```python
df["gap"] = df["actual_productivity"] - df["targeted_productivity"]
df["cumple_meta"] = (df["gap"] >= 0).astype(int)

print(df["cumple_meta"].value_counts(normalize=True).round(3))
print("\nNOTA: la clase mayoritaria domina el dataset, por eso la métrica")
print("primaria del árbol de decisión será F1-macro y no accuracy.")
```

## 3. ¿Qué equipos son el cuello de botella?

```python
resumen = df.groupby("team").agg(
    brecha_promedio=("gap", "mean"),
    tasa_cumplimiento=("cumple_meta", "mean"),
    smv_promedio=("smv", "mean"),
    tiempo_muerto_total=("idle_time", "sum"),
).sort_values("brecha_promedio")

resumen.round(3)
```
